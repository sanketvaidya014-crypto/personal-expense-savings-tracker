# Personal Expense & Savings Tracker
Run cells from top to bottom. This demonstration uses synthetic INR transactions, not your actual finances. First upload the project ZIP.

```python
from google.colab import files
uploaded = files.upload()
```

```python
from pathlib import Path
import zipfile

zip_name = next(name for name in uploaded if name.lower().endswith(".zip"))
with zipfile.ZipFile(zip_name) as z:
    z.extractall("/content/expense_project")
project_folder = next(Path("/content/expense_project").rglob("tracker.py")).parent
print("Project ready:", project_folder)
```

## Run the sample tracker
Each run uses a new output directory. The sample intentionally includes two invalid records.

```python
import subprocess, sys, uuid

subprocess.run([sys.executable, "-m", "pip", "install", "-r", str(project_folder / "requirements.txt")], check=True)
output_folder = project_folder / ("artifacts_" + uuid.uuid4().hex[:8])
subprocess.run([sys.executable, str(project_folder / "tracker.py"), "--input", str(project_folder / "data/sample_transactions.csv"), "--output", str(output_folder)], check=True)
```

```python
import pandas as pd
from IPython.display import display, Markdown

for report in sorted((output_folder / "reports").glob("*.csv")):
    display(Markdown("### " + report.stem.replace("_", " ").title()))
    display(pd.read_csv(report))
```
